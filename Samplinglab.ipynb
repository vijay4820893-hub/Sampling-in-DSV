```python
# Import the libray
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
np.random.seed(42)
df = pd.DataFrame({
    'Student_ID': range(1,101),
    'Name':[f'Student_{i}' for i in range(1,101)],
    'Gender': np.random.choice(['Male', 'Female'],100),
    'Age': np.random.randint(19,24,100),
    'Marks': np.random.randint(40,101,100),
    'Deparment': np.random.choice(['CSE', 'ME', 'ECE', 'CE'],100)
})
```

```python
print(df)
```

```python
df.info()
```

```python
df.describe()
```

```python
# Distribution
print("Distribution of Population(DataSet)\n")
len(df)
#Deparment Distribution
print(df['Deparment'].value_counts())
#Gender Distribution
print(df['Gender'].value_counts())
# Basis Statistics of marks Distribution
print(df['Marks'].describe())
```

```python
# 1. Simple Random Sample: always give an qeual chance to
sample_size = 10
simple_random_sample = df.sample(
    n = sample_size,
    random_state = 42


)
simple_random_sample
```

```python
# Visulalization od Simple Random Sampling
plt.figure(figsize=(12,5))
plt.scatter(df['Student_ID'], df['Marks'], alpha=0.40, label='Population') # Population
plt.scatter(simple_random_sample['Student_ID'],
            simple_random_sample['Marks'],
            s=120, label='SAMPLE')
plt.xlabel("STUDENT ID")
plt.ylabel("MARKS")
plt.title("VISUALIZATION OF SIMPLE RANDOM SAMPLING")
plt.show()

```

```python
# SYSTEMATIC RANDOM SAMPLING
N = len(df)
n = 10
k = N//N
start = np.random.randint(0,k)
systematic_sample = df.iloc[start::k].head(n)
print("Sampling Interval k = ", k)
print("Stsrting Position = ", start)
systematic_sample
```

```python
# Visulalization of systematic Sampling
plt.figure(figsize=(7,4))
plt.scatter(df['Student_ID'], df['Marks'], alpha=0.40, label='Population') # Population
plt.scatter(simple_random_sample['Student_ID'],
            simple_random_sample['Marks'],
            s=120, label='SAMPLE')
plt.xlabel("STUDENT ID")
plt.ylabel("MARKS")
plt.title("VISUALIZATION OF SYSTEMATIC SAMPLING")
plt.show()

```

```python
# 3. Stratified Sample
sample_size = 20
stratified_sample, remaining = train_test_split(
    df, test_size = 0.80,
    stratify = df['Deparment'],
    random_state = 42
)
stratified_sample
print("Original Deparment Proportions:")
print(df['Deparment'].value_counts(normalize=True).round(2))
print("\n Sample Proportions Ratio:")
print(stratified_sample['Deparment'].value_counts(normalize=True).round(3))
```
